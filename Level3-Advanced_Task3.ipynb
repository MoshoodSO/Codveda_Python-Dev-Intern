```python
def solve_n_queens(n):
    # Represent the chessboard as an N x N 2D array initialized with 0
    board = [[0] * n for _ in range(n)]
    solutions = []

    def is_safe(row, col):
        # Check this column for another queen above
        for i in range(row):
            if board[i][col] == 1:
                return False

        # Check upper-left diagonal
        for i, j in zip(range(row - 1, -1, -1), range(col - 1, -1, -1)):
            if board[i][j] == 1:
                return False

        # Check upper-right diagonal
        for i, j in zip(range(row - 1, -1, -1), range(col + 1, n)):
            if board[i][j] == 1:
                return False

        return True

    def backtrack(row):
        # Base case: All queens are placed successfully
        if row == n:
            # Save a deep copy of the current valid board layout
            solutions.append([list(r) for r in board])
            return

        # Try placing a queen in each column of the current row
        for col in range(n):
            if is_safe(row, col):
                board[row][col] = 1 # Place the queen
                
                backtrack(row + 1) # Recur to place the next queen
                
                board[row][col] = 0 # Backtrack and remove the queen

    # Start backtracking from the first row (index 0)
    backtrack(0)
    return solutions
```

```python
def print_solutions(solutions):
    print(f"Found {len(solutions)} total solutions:\n")
    for idx, board in enumerate(solutions, 1):
        print(f"Solution {idx}:")
        for row in board:
            # Print 'Q' for a queen and '.' for an empty space
            print(" ".join("Q" if cell == 1 else "." for cell in row))
        print()
```

```python
# Example usage for a 4x4 chessboard
n = 4
all_solutions = solve_n_queens(n)
print_solutions(all_solutions)
```
