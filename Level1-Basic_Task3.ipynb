```python
def WordCounter(file):
    '''
    A function that count the number of words in a file.

    This function open a file and returns the count of the words in the file.

    Args:
        file (str): The directory of the file

    Return:
        int: the number of the words in the file.

    Raise:
        FileNotFoundError: If the file is not found or the directoryis not correct.

    Example:
    >>> WordCounter("docs.txt")
    62

    '''
    
    try:
        with open(file, 'r') as f:
            content = f.read()
        return len(content.split())

    except FileNotFoundError:
        return f"No such file or directory: {file}"
```

```python
WordCounter("docs.txt")
```
