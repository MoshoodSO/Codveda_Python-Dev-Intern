```python
import requests
```

```python
# Open-Meteo API endpoint
url = "https://api.open-meteo.com/v1/forecast"

# Lagos coordinates
params = {
    "latitude": 6.5244,
    "longitude": 3.3792,
    "current": "temperature_2m,relative_humidity_2m,wind_speed_10m",
    "timezone": "auto"
}
```

```python
try:
    # Make a GET request to the API
    response = requests.get(url, params=params, timeout=10)

    # Raise an exception for HTTP errors
    response.raise_for_status()

    # Convert the response to JSON
    data = response.json()

    # Check that the expected data exists
    if "current" not in data:
        raise ValueError("Invalid API response: current weather data is missing.")

    # Extract current weather information
    current = data["current"]

    temperature = current["temperature_2m"]
    humidity = current["relative_humidity_2m"]
    wind_speed = current["wind_speed_10m"]

    # Display the data in a user-friendly format
    print("\n===== Current Weather =====")
    print("Location: Lagos, Nigeria")
    print(f"Temperature: {temperature} °C")
    print(f"Humidity: {humidity}%")
    print(f"Wind Speed: {wind_speed} km/h")
    print("===========================\n")

except requests.exceptions.Timeout:
    print("Error: The request timed out. Please try again.")

except requests.exceptions.ConnectionError:
    print("Error: Could not connect to the API. Check your internet connection.")

except requests.exceptions.HTTPError as e:
    print(f"Error: The API returned an HTTP error: {e}")

except requests.exceptions.RequestException as e:
    print(f"Error making API request: {e}")

except (ValueError, KeyError) as e:
    print(f"Error processing API response: {e}")
```
